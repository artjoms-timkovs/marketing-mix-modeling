```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Load Data
# In a real scenario, this connects to a Data Warehouse
df = pd.read_csv('data.csv')

# 2. Correlation Matrix (Checking for Multicollinearity)
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix: Media Spend vs Revenue")
plt.show()

# 3. Simple OLS Regression Model
X = df[['TV_Spend', 'Social_Spend', 'Search_Spend', 'Seasonality_Index']]
y = df['Revenue']

model = LinearRegression()
model.fit(X, y)

# 4. Extracting Coefficients (The "True" ROAS)
coef_df = pd.DataFrame(model.coef_, X.columns, columns=['Coefficient (Incremental Value)'])
print("\n--- MMM Model Results: Estimated Incremental Impact ---")
print(coef_df)

print(f"\nBase Revenue (Intercept): {model.intercept_:.2f}")
print(f"Model Accuracy (R-squared): {model.score(X, y):.2f}")
```
